```python
from scipy import stats
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings('ignore')

```

```python
beml_df=pd.read_csv("BEML.csv")
beml_df.head()
```

```python
glaxo_df=pd.read_csv("GLAXO.csv")
glaxo_df.head()
```

```python
beml_df=beml_df[['Date','Close']]
glaxo_df=glaxo_df[['Date','Close']]
```

```python
glaxo_df=glaxo_df.set_index(pd.DatetimeIndex(glaxo_df['Date']))
beml_df=beml_df.set_index(pd.DatetimeIndex(beml_df['Date']))
```

```python
beml_df.head() #now date colums acting like an index
```

```python
glaxo_df.head()   #now date colums acting like an index
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.plot(glaxo_df.Close)
plt.xlabel('Time')
plt.ylabel('Close Price')
```

```python
plt.plot(beml_df.Close)
plt.title('comparasion')
plt.xlabel('Time')
plt.ylabel('Close')
```

```python
glaxo_df['gain'] = glaxo_df.Close.pct_change(periods =1)
beml_df['gain'] = beml_df.Close.pct_change(periods =1)
```

```python
glaxo_df=glaxo_df.dropna()
beml_df =beml_df.dropna()
```

```python
plt.figure(figsize=(8,5))
plt.plot(glaxo_df.gain)
plt.xlabel('Time')
plt.ylabel('gain')
```

```python
plt.figure(figsize=(8,5))
plt.plot(beml_df.gain)
plt.xlabel('Time')
plt.ylabel('gain')
```

```python
glaxo_df
```

```python
plt.figure(figsize=(10,5))
plt.suptitle('Distribution',fontsize=15)
plt.subplot(1,2,1)
sns.distplot(glaxo_df.gain,label='Glaxo')
plt.xlabel('gain')
plt.ylabel('density')
plt.legend()

plt.subplot(1,2,2)
sns.distplot(beml_df.gain,label='BEML')
plt.xlabel('gain')
plt.ylabel('density')
plt.legend()
```

```python
plt.figure(figsize=(10,5))
sns.distplot(glaxo_df.gain,label='Glaxo')
sns.distplot(beml_df.gain,label='BEML')
plt.xlabel('gain')
plt.ylabel('density')
plt.legend()
```

```python
print('Glaxo Mean:',round(glaxo_df.gain.mean(),4))
print('Glaxo Standard Deviation:',round(glaxo_df.gain.std(),4))
```

```python
glaxo_df.gain.mean()
```

```python
print('BEML Mean:',round(beml_df.gain.mean(),4))
print('BEML Standard Deviation:',round(beml_df.gain.std(),4))
```
