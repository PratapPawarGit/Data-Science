**cell center example**

```python
# 1st way
p2=1-stats.t.cdf(1.4,49)
p=p2*2
p
```

```python
# 2nd way
p2=stats.t.cdf(-1.4,49)
p=p2*2
p
```

```python
# 3rd way
p=2*stats.t.cdf(-1.4,49)
print(p)
#print(round(r,4))
print("rounded value",round(p,4))
```
